```python
# Simple List
fruits = ["apple", "banana", "cherry"]
print(fruits)
print(type(fruits))
```

```python
# Creating a list using user input
n = int(input("How many elements? "))
my_list = []

for i in range(n):
    val = input(f"Enter element {i+1}: ")
    my_list.append(val)

print("Final List:", my_list)
```

```python
# Taking all inputs in one line and creating its list
mylist = input("Enter your list items : ").split()
print(mylist)
print(type(mylist))
```

```python
# Taking all inputs in one line and creating its list
mylist = input("Enter your list items : ").split(",") # Separated by => ,
print(mylist)
print(type(mylist))
```

```python
list = ['apple','banana']
list.append('kiwi')
print(list)
list.pop()
print(list)
list.insert(0,'grapes')
print(list)
for fruit in list:
    if(fruit == 'apple'):
        print(f'found it at {list.index(fruit)}')
    else:
        print('not yet')
```

```python
# Creating a tuple
fruits = ("apple", "banana", "cherry")
print(fruits)
```

```python
#User Input for Tuple
n = int(input("How many elements? "))
temp_list = []

for i in range(n):
    val = input(f"Enter element {i+1}: ")
    temp_list.append(val)

my_tuple = tuple(temp_list)
print("Final Tuple:", my_tuple)
print(type(my_tuple))
```

```python
#Dictionary in Python
example = {"name": "Yash Thakur", "age": 23}
print(example)
print(type(example))
```

```python
# Taking User input for dictionary
record = {}

key = input("Enter your key : ")
value = input("Enter your value : ")

record[key] = value

print(record)
print(type(record))
```

```python
# Multiple entries with loop
my_dict = {}
n = int(input("How many items do you have? "))

for i in range(n):
    key = input(f"Enter key {i+1}: ")
    value = input(f"Enter value for {key}: ")
    my_dict[key] = value

print("Final dictionary:", my_dict)
```
