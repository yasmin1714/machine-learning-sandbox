```python
#Conditional Statements
x = 3
if x > 5:
    print("x is greater than 5")
else:
    print("x is not greater than 5")
```

```python
# If-Else & Elif
x = 0

if x > 0:
    print("Positive")
elif x == 0:
    print("Zero")
else:
    print("Negative")
```

```python
# Nested If-Else
x = 15
if x > 0:
    if x % 2 == 0:
        print("Positive Even")
    else:
        print("Positive Odd")
```

```python
# Simple for loop in python
for i in range(5):
    print(i)
```

```python
# Reverse loop in python
for i in range(100, 90, -1):
    print(i)
```

```python
for row in range(0,3):
    for val in range(0,3):
        print(f" [Row: {row}, Col: {val}]", end=" ")
    print()
```

```python
# Looping through a matrix
matrix = [[1,2,3],[4,5,6]]

for row in matrix:
    for val in row:
        print(val, end=" ")
    print()
        
```

```python
# Looping through the list with there indexes
matrix = [[1,2,3],[4,5,6]]

for i, row in enumerate(matrix):
    for j, val in enumerate(row):
        print(f"([{i}][{j}] = {val})", end="  ")
    print("")
        
```

```python
# While loop
start = 1
while (start <= 5):
    print(start)
    start += 1
```

```python
# While loop reverse
count = 5
while (count > 0):
    print(count)
    count -= 1
```

```python
# While loop with CONTINUE
count = 5
while (count > 0):
    if (count == 2):
        # print("We encountered '2' hence skipping it...")
        count -= 1
        continue
    else:
        print(count)
    count -= 1
```

```python
'''
1. Bus Ticket Discount (If-Else)

A bus company gives a 50% discount to children below 12 years.
Write a program that takes age as input and tells if the passenger gets a discount or not.

2. Even & Odd Numbers (Loop + List)

You are given a list of numbers = [12, 7, 9, 20, 33].
Write a program that prints which numbers are Even and which are Odd.

3. Shopping Cart (Tuple)

You bought items stored in a tuple: ("milk", "bread", "eggs", "butter").
Ask the user to enter an item.
If the item is in the tuple → print "Available".
Else → print "Not Available".

4. Student Marks (Dictionary + Loop + If-Else)

Store marks of students in a dictionary:
{"Aman": 76, "Riya": 45, "Kunal": 88}
Write a program to print each student’s name and whether they Passed (marks ≥ 50) or Failed.

5. ATM Cash Withdrawal (While Loop + If-Else)

A person wants to withdraw money from an ATM.
Balance = 5000.
Keep asking the user to enter the withdrawal amount until balance becomes less than 1000.
Print balance after every withdrawal.
'''
```

```python
#1
age = int(input("Enter passenger's age: "))

if age < 12:
    print("Passenger gets 50% discount.")
else:
    print("No discount.")
```

```python
#2
numbers = [12, 7, 9, 20, 33]

for num in numbers:
    if num % 2 == 0:
        print(f"{num} is Even")
    else:
        print(f"{num} is Odd")
```

```python
#3
items = ("milk", "bread", "eggs", "butter")

search = input("Enter item to search: ").lower()

if search in items:
    print(f"{search} is Available")
else:
    print(f"{search} is Not Available")
```

```python
marks = {"Aman": 76, "Riya": 45, "Kunal": 88}
# print(marks.items())
for name, score in marks.items():
    if score >= 50:
        print(f"{name} Passed with {score} marks")
    else:
        print(f"{name} Failed with {score} marks")
```

```python
balance = 5000

while balance >= 1000:
    amount = int(input("Enter amount to withdraw: "))

    if amount <= balance:
        balance -= amount
        print(f"Withdrawal successful! Remaining balance: {balance}")
    else:
        print("Insufficient balance!")

print("Balance is less than 1000. Cannot withdraw further.")
```
